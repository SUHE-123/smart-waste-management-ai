```python
from ultralytics import YOLO

# Load YOLOv8 Nano
model = YOLO("yolov8n.pt")

# Training final
results = model.train(
    data="/kaggle/working/data.yaml",

    epochs=50,
    imgsz=640,
    batch=16,

    device=0,

    # checkpoint
    save=True,
    save_period=5,

    # cache dataset
    cache=True
)
```

```python
from ultralytics import YOLO
import cv2

# Load model
model = YOLO("best.pt")

# Override nama class model
model.names = {
    17: "daun",
    18: "kaca",
    19: "kaleng",
    20: "kardus",
    21: "kertas",
    22: "plastik",
    23: "sedotan"
}

# Webcam
cap = cv2.VideoCapture(0)

while True:

    ret, frame = cap.read()

    if not ret:
        break

    # Prediksi
    results = model(frame)

    # Render otomatis YOLO
    annotated_frame = results[0].plot()

    # Tampilkan
    cv2.imshow("Smart Waste AI Detection", annotated_frame)

    # Tombol keluar
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()
```
